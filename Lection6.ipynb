Перетащите компонент LED на плату ESP32.
Подключите один из выводов светодиода к GPIO-пину ESP32 (например, к пину GPIO 5).
Пример кода для управления светодиодом:

```python
import machine
import time

# Создание объекта для пина 5
led = machine.Pin(5, machine.Pin.OUT)

# Включение и выключение светодиода
while True:
    led.value(1)  # Включить
    time.sleep(1)
    led.value(0)  # Выключить
    time.sleep(1)
```

Этот код включает и выключает светодиод каждую секунду.

Этот код заставит робота двигаться вперед, назад и поворачивать влево/вправо.

### Задание 2: Управление роботом с двумя моторами и датчиком

**Цель:** Создайте программу для робота, который двигается вперед, пока не обнаружит препятствие, а затем останавливается.

**Описание:**

Используйте ESP32 в симуляторе Wokwi.
Подключите два мотора и ультразвуковой датчик.
Напишите программу, которая будет двигать робота вперед, пока датчик не обнаружит препятствие на расстоянии менее 20 см.
После обнаружения препятствия робот должен остановиться.

**Дополнительно:**

После остановки робот должен сделать несколько поворотов на 90 градусов и затем снова двигаться вперед, если препятствий нет.

```python
# Блок для решения задания 2
from machine import Pin, time_pulse_us
from time import sleep_us, sleep_ms, sleep

TRIG_PIN = 5
ECHO_PIN = 4

DIR1_PIN = 17
STEP1_PIN = 16

DIR2_PIN = 19
STEP2_PIN = 18

trig = Pin(TRIG_PIN, Pin.OUT)
echo = Pin(ECHO_PIN, Pin.IN)

dir1 = Pin(DIR1_PIN, Pin.OUT)
step1 = Pin(STEP1_PIN, Pin.OUT)

dir2 = Pin(DIR2_PIN, Pin.OUT)
step2 = Pin(STEP2_PIN, Pin.OUT)

def get_distance():
    trig.value(0)
    sleep_us(2)
    trig.value(1)
    sleep_us(10)
    trig.value(0)
    duration = time_pulse_us(echo, 1, 30000)
    distance = (duration / 2) / 29.1
    return distance

def step_both():
    step1.value(1)
    step2.value(1)
    sleep_us(800)
    step1.value(0)
    step2.value(0)
    sleep_us(800)

def step_turn():
    step1.value(1)
    step2.value(1)
    sleep_us(800)
    step1.value(0)
    step2.value(0)
    sleep_us(800)

def move_forward(steps=100):
    dir1.value(1)
    dir2.value(1)
    for _ in range(steps):
        step_both()

def stop():
    sleep(0.5)

def turn_90():
    dir1.value(0)
    dir2.value(1)
    for _ in range(100):
        step_turn()
    stop()

def main():
    while True:
        dist = get_distance()
        print("Расстояние:", dist, "см")

        if dist < 20:
            print("Препятствие! Остановка.")
            stop()
            turn_90()
            sleep(0.5)
        else:
            move_forward(50)

        sleep_ms(50)

main()
```

**Цель:** Разработайте программу для робота, который будет двигаться по комнате, избегать препятствий и искать свободное пространство.

**Описание:**

Используйте ESP32 в симуляторе Wokwi.
Подключите два мотора и ультразвуковой датчик.
Напишите программу, которая:
Начинает движение вперед.
При обнаружении препятствия поворачивает на 90 градусов и продолжает движение в новом направлении.
Если после поворота препятствие снова обнаружено, программа должна снова повернуть на 90 градусов в другую сторону.
Программа должна продолжать искать свободное пространство, пока не найдется место для свободного движения.

**Дополнительно:**

Используйте несколько датчиков, если возможно, для улучшения результатов поиска свободного пространства.


```python
# Блок для решения задания 4
from machine import Pin, time_pulse_us
from time import sleep_us, sleep_ms, sleep

TRIG_F = 5
ECHO_F = 4

TRIG_L = 23
ECHO_L = 22

TRIG_R = 27
ECHO_R = 26

DIR1_PIN = 17
STEP1_PIN = 16
DIR2_PIN = 19
STEP2_PIN = 18

sensors = {
    "front": {"trig": Pin(TRIG_F, Pin.OUT), "echo": Pin(ECHO_F, Pin.IN)},
    "left":  {"trig": Pin(TRIG_L, Pin.OUT), "echo": Pin(ECHO_L, Pin.IN)},
    "right": {"trig": Pin(TRIG_R, Pin.OUT), "echo": Pin(ECHO_R, Pin.IN)}
}

dir1 = Pin(DIR1_PIN, Pin.OUT)
step1 = Pin(STEP1_PIN, Pin.OUT)
dir2 = Pin(DIR2_PIN, Pin.OUT)
step2 = Pin(STEP2_PIN, Pin.OUT)

def get_distance(sensor):
    trig = sensors[sensor]["trig"]
    echo = sensors[sensor]["echo"]
    trig.value(0)
    sleep_us(2)
    trig.value(1)
    sleep_us(10)
    trig.value(0)
    duration = time_pulse_us(echo, 1, 30000)
    if duration <= 0:
        return 999
    distance = (duration / 2) / 29.1
    return distance

def step_both(delay_us=500):
    step1.value(1)
    step2.value(1)
    sleep_us(delay_us)
    step1.value(0)
    step2.value(0)
    sleep_us(delay_us)

def step_turn(delay_us=500):
    step1.value(1)
    step2.value(1)
    sleep_us(delay_us)
    step1.value(0)
    step2.value(0)
    sleep_us(delay_us)

def move_forward(steps=100):
    dir1.value(1)
    dir2.value(1)
    for _ in range(steps):
        step_both()

def stop():
    sleep(0.3)

def turn(direction="right"):
    if direction == "right":
        dir1.value(0)
        dir2.value(1)
        steps = 100
    elif direction == "left":
        dir1.value(1)
        dir2.value(0)
        steps = 100
    elif direction == "back":
        dir1.value(0)
        dir2.value(1)
        steps = 200
    else:
        return

    for _ in range(steps):
        step_turn()
    stop()

def main():

    while True:
        dist_f = get_distance("front")
        dist_l = get_distance("left")
        dist_r = get_distance("right")

        print("Дистанции (см):  перед = {:.1f},  лево = {:.1f},  право = {:.1f}".format(
            dist_f, dist_l, dist_r))

        if dist_f < 20:
            print("Препятствие! Остановка.")
            stop()

            if dist_l > dist_r and dist_l > 20:
                print("Поворачиваем налево.")
                turn("left")
            elif dist_r > dist_l and dist_r > 20:
                print("Поворачиваем направо.")
                turn("right")
            else:
                print("Обе стороны заняты. Разворот на 180°.")
                turn("back")

            sleep(0.5)
        else:
            move_forward(50)

        sleep_ms(50)

main()
```

После этого уменьшите скорость обратно до 0 за 5 секунд.

**Требования:**

Используйте функцию machine.PWM для управления скоростью моторов.
Убедитесь, что скорость увеличивается и уменьшается плавно.

```python
# Блок для решения задания 5
from machine import Pin, PWM
from time import sleep

DIR1_PIN = 17
STEP1_PIN = 16

DIR2_PIN = 19
STEP2_PIN = 18

dir1 = Pin(DIR1_PIN, Pin.OUT)
dir2 = Pin(DIR2_PIN, Pin.OUT)

dir1.value(1)
dir2.value(1)

step1_pwm = PWM(Pin(STEP1_PIN))
step2_pwm = PWM(Pin(STEP2_PIN))

step1_pwm.freq(1)
step2_pwm.freq(1)
step1_pwm.duty(512)
step2_pwm.duty(512)


for f in range(1, 1001, 10):  # от 1 Гц до 1000 Гц за 5 секунд
    step1_pwm.freq(f)
    step2_pwm.freq(f)
    sleep(0.05)  # 0.05 * 100 шагов = 5 секунд
print("Максимальная скорость")

for f in range(1000, 0, -10):  # обратно до 0
    step1_pwm.freq(f)
    step2_pwm.freq(f)
    sleep(0.05)
print("Остановка")

step1_pwm.deinit()
step2_pwm.deinit()
```
