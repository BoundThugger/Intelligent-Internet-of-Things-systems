# Лекция 5. Использование ООП в Python на примере с моделью робота

Предположим, мы создаём программу для моделирования робота с использованием ООП. Для этого определим несколько классов, которые будут описывать различные компоненты робота, такие как **Робот**, **Двигатель**, **Колесо** и т. д.

## 5.1 Класс `Motor` (Двигатель)

Для начала создадим класс, который будет описывать двигатель робота. Он будет иметь атрибуты мощности и максимальной скорости.

```python
class Motor:
    def __init__(self, max_speed, power):
        self.max_speed = max_speed  # Максимальная скорость мотора
        self.power = power  # Мощность мотора

    def set_speed(self, speed):
        """Устанавливаем скорость мотора"""
        if 0 <= abs(speed) <= self.max_speed:
            self.speed = speed
        else:
            print("Скорость вне диапазона!")

    def get_speed(self):
        """Получаем текущую скорость мотора"""
        return self.speed
```

## 5.2 Класс `Wheel` (Колесо)

Каждое колесо робота может вращаться и влиять на его движение. Мы создадим класс для колеса, который будет иметь атрибуты радиуса и текущей скорости вращения.

```python
class Wheel:
    def __init__(self, radius):
        self.radius = radius  # Радиус колеса
        self.rotation_speed = 0  # Скорость вращения колеса

    def rotate(self, speed):
        """Устанавливаем скорость вращения колеса"""
        self.rotation_speed = speed

    def get_rotation_speed(self):
        """Получаем текущую скорость вращения колеса"""
        return self.rotation_speed
```

## 5.3 Класс `Robot` (Робот)

Теперь мы создадим класс робота, который будет включать в себя два двигателя и два колеса. Робот будет управляться через методы для движения, поворота и остановки.

```python
class Robot:
    def __init__(self, motor_left, motor_right, wheel_left, wheel_right):
        self.motor_left = motor_left  # Левый двигатель
        self.motor_right = motor_right  # Правый двигатель
        self.wheel_left = wheel_left  # Левое колесо
        self.wheel_right = wheel_right  # Правое колесо

    def move_forward(self, speed):
        """Движение робота вперед"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(speed)

    def turn_left(self, speed):
        """Поворот налево"""
        self.motor_left.set_speed(0)  # Левый мотор не вращается
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(speed)

    def turn_right(self, speed):
        """Поворот направо"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(0)  # Правый мотор не вращается
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(0)

    def stop(self):
        """Остановка робота"""
        self.motor_left.set_speed(0)
        self.motor_right.set_speed(0)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(0)
```

## 5.4 Пример использования

```python
# Создаем двигатели и колеса
motor_left = Motor(max_speed=10, power=100)
motor_right = Motor(max_speed=10, power=100)
wheel_left = Wheel(radius=5)
wheel_right = Wheel(radius=5)

# Создаем робота
robot = Robot(motor_left, motor_right, wheel_left, wheel_right)

# Двигаем робота вперед с определенной скоростью
robot.move_forward(5)
print(f"Скорость левого мотора: {motor_left.get_speed()} м/с")
print(f"Скорость правого мотора: {motor_right.get_speed()} м/с")

# Поворот налево
robot.turn_left(5)
```

В этом примере мы создали объект робота, который может двигаться вперед и поворачивать налево или направо, управляя двигателями и колесами с помощью методов. Классы `Motor` и `Wheel` инкапсулируют логику работы двигателей и колес, позволяя роботу быть более гибким и расширяемым.

## 5.5 Расчёт скорости передвижения колесного робота

Теперь, когда мы построили структуру для робота, давайте добавим расчет скорости передвижения робота. Для этого нужно учитывать два момента:

1. Скорость вращения колес: Это скорость, с которой вращается колесо робота.
2. Радиус колес: Важно учитывать, что колесо будет двигаться по окружности.

### 5.5.1 Формула для расчета линейной скорости

Скорость робота (линейная скорость) зависит от скорости вращения колес и их радиуса. Формула для расчета линейной скорости выглядит следующим образом:

$v = \omega * r$

где:
- $v$ - линейная скорость робота (м/с)
- $\omega$ - угловая скорость колес (об/с)
- r - радиус колеса (м).

Для расчета скорости робота, можно использовать текущую скорость вращения колес и их радиус:

```python
class Robot:
    def __init__(self, motor_left, motor_right, wheel_left, wheel_right):
        self.motor_left = motor_left
        self.motor_right = motor_right
        self.wheel_left = wheel_left
        self.wheel_right = wheel_right

    def calculate_speed(self):
        """Расчет линейной скорости робота"""
        # Получаем скорость вращения колес
        left_wheel_speed = self.wheel_left.get_rotation_speed()
        right_wheel_speed = self.wheel_right.get_rotation_speed()

        # Считаем среднюю скорость колес
        average_wheel_speed = (left_wheel_speed + right_wheel_speed) / 2

        # Расчет линейной скорости робота
        speed = average_wheel_speed * self.wheel_left.radius
        return speed

    def move_forward(self, speed):
        """Движение робота вперед"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(speed)

    def turn_left(self, speed):
        """Поворот налево"""
        self.motor_left.set_speed(0)
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(speed)

    def turn_right(self, speed):
        """Поворот направо"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(0)
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(0)

    def stop(self):
        """Остановка робота"""
        self.motor_left.set_speed(0)
        self.motor_right.set_speed(0)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(0)
```

## 5.5.2 Пример расчета скорости

```python
# Создаем объекты робота
motor_left = Motor(max_speed=10, power=100)
motor_right = Motor(max_speed=10, power=100)
wheel_left = Wheel(radius=0.1)  # Радиус колеса 10 см
wheel_right = Wheel(radius=0.1)

robot = Robot(motor_left, motor_right, wheel_left, wheel_right)

# Двигаем робота вперед
robot.move_forward(5)

# Рассчитываем скорость робота
speed = robot.calculate_speed()
print(f"Линейная скорость робота: {speed:.2f} м/с")
```

## Пояснение:

- Мы создаем два колеса с радиусом 0.1 метра (10 см).
- Рассчитываем скорость робота, используя формулу $v = \omega * r$, где - $\omega$ - скорость вращения колес, и r - радиус колеса (м).
- Метод `calculate_speed()` учитывает среднюю скорость вращения обоих колес для вычисления общей линейной скорости.

## Задание 2: Добавление скорости поворота робота

Добавьте возможность для робота поворачиваться с заданной угловой скоростью. Например, если оба мотора вращаются с одинаковой скоростью, робот будет двигаться по прямой. Если скорости моторов различны, робот будет поворачиваться.

Создайте метод `turn_with_speed()`, который позволяет роботу поворачиваться с заданной угловой скоростью.
Убедитесь, что робот поворачивает, изменяя скорости двигателей.


```python
# Код для решения задания 2
class Robot:
    def __init__(self, motor_left, motor_right, wheel_left, wheel_right, wheel_base):
        self.motor_left = motor_left
        self.motor_right = motor_right
        self.wheel_left = wheel_left
        self.wheel_right = wheel_right
        self.wheel_base = wheel_base

    def calculate_speed(self):
        """Расчет линейной скорости робота"""
        # Получаем скорость вращения колес
        left_wheel_speed = self.wheel_left.get_rotation_speed()
        right_wheel_speed = self.wheel_right.get_rotation_speed()

        v_left = left_wheel_speed * self.wheel_left.radius
        v_right = right_wheel_speed * self.wheel_right.radius

        # Считаем среднюю скорость колес
        average_wheel_speed = (v_left + v_right) / 2

        # Расчет линейной скорости робота
        speed = average_wheel_speed * self.wheel_left.radius
        return speed

    def turn_with_speed(self, turn_speed, v_center=0):
        # Вычисляем линейные скорости каждого колеса
        v_left = v_center - (turn_speed * self.wheel_base / 2)
        v_right = v_center + (turn_speed * self.wheel_base / 2)

        # Переводим в угловые скорости колёс
        omega_left = v_left / self.wheel_left.radius
        omega_right = v_right / self.wheel_right.radius

        # Применяем к моторам и колёсам
        self.motor_left.set_speed(omega_left)
        self.motor_right.set_speed(omega_right)
        self.wheel_left.rotate(omega_left)
        self.wheel_right.rotate(omega_right)

    def move_forward(self, speed):
        """Движение робота вперед"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(speed)

    def turn_left(self, speed):
        """Поворот налево"""
        self.motor_left.set_speed(0)
        self.motor_right.set_speed(speed)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(speed)

    def turn_right(self, speed):
        """Поворот направо"""
        self.motor_left.set_speed(speed)
        self.motor_right.set_speed(0)
        self.wheel_left.rotate(speed)
        self.wheel_right.rotate(0)

    def stop(self):
        """Остановка робота"""
        self.motor_left.set_speed(0)
        self.motor_right.set_speed(0)
        self.wheel_left.rotate(0)
        self.wheel_right.rotate(0)

motor_left = Motor(max_speed=10, power=100)
motor_right = Motor(max_speed=10, power=100)
wheel_left = Wheel(radius=0.05)
wheel_right = Wheel(radius=0.05)

robot = Robot(motor_left, motor_right, wheel_left, wheel_right, wheel_base=0.2)

robot.turn_with_speed(turn_speed=1)

print("Левое колесо:", wheel_left.get_rotation_speed(), "рад/с")
print("Правое колесо:", wheel_right.get_rotation_speed(), "рад/с")
```

## Задание 3: Создание робота с двумя независимыми двигателями

Создайте нового робота с двумя независимыми двигателями. Каждый двигатель должен быть независимым и управляться своим собственным пультом. Например, один пульт управляет левым колесом, а другой — правым.

1. Реализуйте два пульта для управления каждым мотором.
2. Протестируйте робота, двигая его вперед, назад и поворачивая влево и вправо, управляя каждым мотором независимо.


```python
# Код для решения задания 3
class Controller:
    def __init__(self, motor, wheel):
        self.motor = motor
        self.wheel = wheel

    def set_speed(self, speed):
        self.motor.set_speed(speed)
        self.wheel.rotate(speed)

    def stop(self):
        self.motor.set_speed(0)
        self.wheel.rotate(0)


class RobotDualControl:
    def __init__(self, controller_left, controller_right, wheel_base):
        self.controller_left = controller_left
        self.controller_right = controller_right
        self.wheel_base = wheel_base  # Расстояние между колёсами

    def move_forward(self, speed):
        self.controller_left.set_speed(speed)
        self.controller_right.set_speed(speed)

    def move_backward(self, speed):
        self.controller_left.set_speed(-speed)
        self.controller_right.set_speed(-speed)

    def turn_left(self, speed):
        self.controller_left.set_speed(0)
        self.controller_right.set_speed(speed)

    def turn_right(self, speed):
        self.controller_left.set_speed(speed)
        self.controller_right.set_speed(0)

    def rotate_in_place(self, angular_speed):
        self.controller_left.set_speed(-angular_speed)
        self.controller_right.set_speed(angular_speed)

    def stop(self):
        self.controller_left.stop()
        self.controller_right.stop()

motor_left = Motor(max_speed=10, power=100)
motor_right = Motor(max_speed=10, power=100)
wheel_left = Wheel(radius=0.05)
wheel_right = Wheel(radius=0.05)

controller_left = Controller(motor_left, wheel_left)
controller_right = Controller(motor_right, wheel_right)

robot = RobotDualControl(controller_left, controller_right, wheel_base=0.2)

robot.move_forward(5)
print("Вперед:")
print("Левое колесо:", wheel_left.get_rotation_speed(), "м/с")
print("Правое колесо:", wheel_right.get_rotation_speed(), "м/с")

robot.turn_right(5)
print("\nПоворот направо:")
print("Левое колесо:", wheel_left.get_rotation_speed(), "м/с")
print("Правое колесо:", wheel_right.get_rotation_speed(), "м/с")

robot.rotate_in_place(3)
print("\nРазворот на месте:")
print("Левое колесо:", wheel_left.get_rotation_speed(), "м/с")
print("Правое колесо:", wheel_right.get_rotation_speed(), "м/с")

robot.stop()
print("\nПосле остановки:")
print("Левое колесо:", wheel_left.get_rotation_speed(), "м/с")
print("Правое колесо:", wheel_right.get_rotation_speed(), "м/с")
```

```python
# Код для решения задания 5
import math
import random

class Motor:
    def __init__(self, max_speed, power):
        self.max_speed = max_speed
        self.power = power
        self.current_speed = 0

    def set_speed(self, speed):
        if speed > self.max_speed:
            speed = self.max_speed
        elif speed < -self.max_speed:
            speed = -self.max_speed
        self.current_speed = speed

    def get_speed(self):
        return self.current_speed

    def stop(self):
        self.current_speed = 0


class Wheel:
    def __init__(self, radius):
        self.radius = radius
        self.angular_speed = 0

    def rotate(self, speed):
        self.angular_speed = speed

    def get_rotation_speed(self):
        return self.angular_speed


class Controller:
    def __init__(self, motor, wheel):
        self.motor = motor
        self.wheel = wheel

    def set_speed(self, speed):
        self.motor.set_speed(speed)
        self.wheel.rotate(speed)

    def get_wheel_speed(self):
        return self.wheel.get_rotation_speed() * self.wheel.radius

    def stop(self):
        self.motor.stop()
        self.wheel.rotate(0)


class Sensor:
    def __init__(self, max_distance=5.0):
        self.max_distance = max_distance
        self.distance = max_distance

    def measure(self, robot_x, robot_y, robot_angle, obstacles):
        min_dist = self.max_distance
        for ox, oy in obstacles:
            dx = ox - robot_x
            dy = oy - robot_y
            dist = math.sqrt(dx**2 + dy**2)
            angle_diff = abs(math.atan2(dy, dx) - robot_angle)
            if angle_diff < math.pi/4 and dist < min_dist:
                min_dist = dist
        self.distance = min_dist
        return self.distance

    def is_obstacle_near(self, threshold=1.0):
        return self.distance < threshold


class RobotDualControl:
    def __init__(self, controller_left, controller_right, wheel_base, sensors=None):
        self.controller_left = controller_left
        self.controller_right = controller_right
        self.wheel_base = wheel_base
        self.sensors = sensors if sensors else []
        self.x = 0.0
        self.y = 0.0
        self.angle = 0.0

    def move_forward(self, speed):
        self.controller_left.set_speed(speed)
        self.controller_right.set_speed(speed)

    def stop(self):
        self.controller_left.stop()
        self.controller_right.stop()

    def update_position(self, dt=0.2):
        # линейные скорости колёс
        v_l = self.controller_left.get_wheel_speed()
        v_r = self.controller_right.get_wheel_speed()
        v = (v_l + v_r) / 2
        omega = (v_r - v_l) / self.wheel_base

        self.angle += omega * dt
        self.x += v * math.cos(self.angle) * dt
        self.y += v * math.sin(self.angle) * dt

    def avoid_obstacles(self, obstacles):
        for sensor in self.sensors:
            dist = sensor.measure(self.x, self.y, self.angle, obstacles)
            if sensor.is_obstacle_near(threshold=0.5):
                print("Обнаружено препятствие! Изменение направления движения")
                self.turn_random()
                return True
        return False

    def turn_random(self):
        turn_angle = random.uniform(-math.pi/2, math.pi/2)
        self.angle += turn_angle
        print(f"Поворот на угол {math.degrees(turn_angle):.1f}°")

    def step(self, speed, obstacles, dt=0.2):
        if not self.avoid_obstacles(obstacles):
            self.move_forward(speed)
        else:
            self.stop()
        self.update_position(dt)


class Resistance:
    def __init__(self, max_factor=1.0):
        self.factor = 0.0
        self.max_factor = max_factor

    def set_factor(self, value):
        if value < 0:
            value = 0
        elif value > self.max_factor:
            value = self.max_factor
        self.factor = value

    def apply(self, speed):
        return speed * (1.0 - self.factor)


class RobotDualControlWithResistance(RobotDualControl):
    def __init__(self, controller_left, controller_right, wheel_base, sensors=None, resistance=None):
        super().__init__(controller_left, controller_right, wheel_base, sensors)
        self.resistance = resistance if resistance else Resistance()

    def move_forward(self, speed):
        adjusted_speed = self.resistance.apply(speed)
        self.controller_left.set_speed(adjusted_speed)
        self.controller_right.set_speed(adjusted_speed)




if __name__ == "__main__":
    motor_left = Motor(max_speed=10, power=100)
    motor_right = Motor(max_speed=10, power=100)
    wheel_left = Wheel(radius=0.05)
    wheel_right = Wheel(radius=0.05)
    controller_left = Controller(motor_left, wheel_left)
    controller_right = Controller(motor_right, wheel_right)
    front_sensor = Sensor(max_distance=5.0)
    resistance = Resistance(max_factor=0.5)

    robot = RobotDualControlWithResistance(controller_left, controller_right, wheel_base=0.2,
                                           sensors=[front_sensor], resistance=resistance)

    # Препятствия
    obstacles = [(0.8, 0.0), (1.0, 0.5)]

    for t in range(30):
        resistance.set_factor(t * 0.02)
        robot.step(speed=5.0, obstacles=obstacles, dt=0.2)
        print(f"t={t:2d}: позиция=({robot.x:.2f}, {robot.y:.2f}), угол={math.degrees(robot.angle):.1f}°, расстояние={front_sensor.distance:.2f}")
```
