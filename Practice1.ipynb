## Краткая справка по функции input() в Python
Функция input() используется для получения ввода от пользователя через консоль. Она приостанавливает выполнение программы и ждет, пока пользователь введет данные и нажмет Enter.

Синтаксис:

```python
input([prompt])
```

prompt (необязательный аргумент): строка, которая выводится на экран перед ожиданием ввода. Это может быть подсказка для пользователя.

Возвращаемое значение:
Функция всегда возвращает введенные данные в виде строки (тип str).

Примеры использования:

### 1. Простой ввод:

```python
name = input("Введите ваше имя: ")
print(f"Привет, {name}!")
```

### 2.Преобразование ввода:
Поскольку input() возвращает строку, для работы с числами нужно преобразовать ввод в нужный тип (например, int или float).

```python
age = int(input("Введите ваш возраст: "))
print(f"Через 5 лет вам будет {age + 5} лет.")
```

### 3.Ввод нескольких значений:
Можно использовать split() для разделения ввода на несколько значений.

```python
x, y = input("Введите два числа через пробел: ").split()
x = int(x)
y = int(y)
print(f"Сумма: {x + y}")
```

### 4.Обработка ошибок:
Если пользователь вводит некорректные данные (например, буквы вместо чисел), можно использовать обработку исключений.

```python
try:
    number = int(input("Введите число: "))
    print(f"Вы ввели: {number}")
except ValueError:
    print("Ошибка: введено не число!")
```

Важные моменты:
Ввод всегда возвращается как строка, даже если пользователь вводит число.

Если нужно работать с числами, используйте функции преобразования: int(), float() и т.д.

Для ввода нескольких значений можно использовать split() или несколько вызовов input().

Функция input() — это простой и удобный способ взаимодействия с пользователем в консольных приложениях. При решении задач ниже вы должны использовать эту функцию, так что можете обращаться к данной справке за помощью.

## Задача 1
***
Напишите программу, которая симулирует движение робота вперед на заданное расстояние. Используйте переменные для хранения скорости и времени движения, а затем вычислите пройденное расстояние по формуле:
расстояние = скорость * время.

```python
#Напишите ваше решение в этом блоке и запустите его
flag_v = True
flag_t = True
while flag_v:
  try:
    v = abs(float(input('Введите скорость робота в м/с: ')))
    flag_v = False
  except ValueError:
    print('Введите корректное значение скорости!')

while flag_t:
  try:
    t = float(input('Введите время движения в с: '))
    if t >= 0:
      flag_t = False
    else:
      print('Время не может быть отрицательным!')
  except ValueError:
    print('Введите корректное значение времени!')
print(f'За {t} с со скоростью {v} м/с робот прошел {v*t} м')
```

## Задача 4
***
Напишите программу, которая симулирует движение робота по квадратной траектории. Робот должен двигаться вперед, поворачивать на 90 градусов и повторять это 4 раза.

```python
import math

class Robot:
    def __init__(self, wheel_radius=0.05, wheel_base=0.2):
        self.wheel_radius = wheel_radius
        self.wheel_base = wheel_base
        self.x = 0.0
        self.y = 0.0
        self.angle = 0.0
        self.time = 0.0

    def move_forward(self, speed, duration, dt=0.1):
        for _ in range(int(duration / dt)):
            v = speed * self.wheel_radius
            self.x += v * math.cos(self.angle) * dt
            self.y += v * math.sin(self.angle) * dt
            self.time += dt

    def turn_right(self):
        self.angle += math.pi / 2
        self.angle %= 2 * math.pi
        self.time += 1.0

    def print_state(self, step_name):
        print(f"{step_name}: позиция=({abs(self.x):.2f}, {abs(self.y):.2f}), угол={math.degrees(self.angle):.1f}°, время={self.time:.1f} c")

if __name__ == "__main__":
    robot = Robot()

    linear_speed = 10.0
    side_time = 2.0

    print("Робот движется по квадратной траектории:\n")

    robot.print_state("Шаг 0 — начальное положение")

    for i in range(4):
        robot.move_forward(linear_speed, side_time)
        robot.print_state(f"Шаг {i*2+1} — движение вперед")

        robot.turn_right()
        robot.print_state(f"Шаг {i*2+2} — поворот на 90°")
    print(f"\nОбщее время движения: {robot.time:.1f} секунд")

```

## Задача 5
***
Робот работает от батареи. Напишите программу, которая вычисляет оставшийся заряд батареи после выполнения задачи. Изначальный заряд батареи — 100%, а каждая задача съедает 10% заряда.

```python
#Напишите ваше решение в этом блоке и запустите его
import math

class Robot:
    def __init__(self, wheel_radius=0.05, wheel_base=0.2):
        self.wheel_radius = wheel_radius
        self.wheel_base = wheel_base
        self.x = 0.0
        self.y = 0.0
        self.angle = 0.0
        self.time = 0.0
        self.battery = 100.0

    def down_battery(self, amount=10.0):
        self.battery -= amount
        if self.battery < 0:
            self.battery = 0

    def move_forward(self, speed, duration, dt=0.1):
        for _ in range(int(duration / dt)):
            v = speed * self.wheel_radius
            self.x += v * math.cos(self.angle) * dt
            self.y += v * math.sin(self.angle) * dt
            self.time += dt
        self.down_battery()

    def turn_right(self):
        self.angle += math.pi / 2
        self.angle %= 2 * math.pi
        self.time += 1.0
        self.down_battery()

    def print_state(self, step_name):
        print(f"{step_name}: позиция=({abs(self.x):.2f}, {abs(self.y):.2f}), угол={math.degrees(self.angle):.1f}°, "
              f"время={self.time:.1f} c, заряд={self.battery:.1f}%")


if __name__ == "__main__":
    robot = Robot()

    linear_speed = 10.0
    side_time = 2.0

    print("Движение по квадратной траектории:\n")

    robot.print_state("Шаг 0 — начальное положение")

    for i in range(4):
        robot.move_forward(linear_speed, side_time)
        robot.print_state(f"Шаг {i*2+1} — движение вперед")

        robot.turn_right()
        robot.print_state(f"Шаг {i*2+2} — поворот на 90°")

    print(f"\nОбщее время движения: {robot.time:.1f} секунд")
    print(f"Остаток заряда батареи: {robot.battery:.1f}%")

```
